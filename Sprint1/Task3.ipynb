Описание задачи:

Создать базовый класс CreditCard в C#, который будет представлять информацию о кредитных картах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу CreditCard:

• Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок действия (ExpiryDate).

• Методы:

o GetInfo(): метод для получения информации о кредитной карте в виде строки.

o Pay(): метод для оплаты покупки с использованием карты.

o CheckBalance(): метод для проверки баланса на карте.

Требования к производным классам:

GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты, такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть переопределен для добавления информации о получении бонусных миль при оплате покупки.
PremiumCard (PremiumCreditCard): Должен содержать дополнительные атрибуты, такие как Ассистент поддержки (SupportAssistant). Метод CheckBalance() должен быть переопределен для предоставления возможности обратиться за помощью к ассистенту поддержки в случае проблем с балансом.
CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Компания (Company). Метод GetInfo() должен быть переопределен для включения информации о компании в описании карты.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

```
public interface ICreditCard 
{
    int CardNumber { get; set; }
    string HolderName { get; set; }
    int ExpiryDate { get; set; }

    void GetInfo();
    void Pay(int payment);
    void CheckBalance();
}

public interface IGoldCreditCard : ICreditCard 
{
    int BonusMiles { get; set; }
}

public interface IPremiumCreditCard : ICreditCard 
{
    string SupportAssistant { get; set; }
}

public interface ICorporateCreditCard : ICreditCard
{
    string Company { get; set; }
}

public class CreditCard : ICreditCard 
{
    public int CardNumber { get; set; }
    public string HolderName { get; set; }
    public int ExpiryDate { get; set; }
    protected int _balance;

    public CreditCard(int cardNumber, string holderName, int expiryDate, int startBalance) 
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        _balance = startBalance;
    }

    public virtual void GetInfo() 
    {
        Console.WriteLine($"Номер кредитной карты: {CardNumber}; Владелец: {HolderName}; Срок действия: {ExpiryDate}.");
    }

    public virtual void Pay(int payment) 
    {
        if (payment > _balance)
        {
            Console.WriteLine($"Ваш баланс отрицательный! Баланс: {_balance}");
            return;
        }
        
        _balance = _balance - payment;
        Console.WriteLine($"Оплата покупки на {payment} рублей. Остаток: {_balance}");
    }

    public virtual void CheckBalance() 
    {
        Console.WriteLine($"Баланс кредитной карты: {_balance} рублей");
    }
}

public class GoldCreditCard : CreditCard, IGoldCreditCard 
{
    public int BonusMiles { get; set; }    

    public GoldCreditCard(int cardNumber, string holderName, int expiryDate, int startBalance, int bonusMiles) : base (cardNumber, holderName, expiryDate, startBalance)
    {
        BonusMiles = bonusMiles;
    }

    public override void Pay(int payment) 
    {
        if (payment > _balance)  
        {
            Console.WriteLine("Недостаточно средств!");
            return;
        }
        
        _balance -= payment;
        
        int earnedMiles = payment >= 10000 ? 2 : 0;
        BonusMiles += earnedMiles;

        if (payment >= 10000) 
        {
            Console.WriteLine($"Покупка на {payment} рублей! Начислено миль: +{earnedMiles}");
            Console.WriteLine($"Всего миль: {BonusMiles}");
        }
        else
        {
            Console.WriteLine($"Покупка на {payment} рублей. Миль не начислено.");
        }
    }
}

public class PremiumCreditCard : CreditCard, IPremiumCreditCard
{
    public string SupportAssistant { get; set;}

    public PremiumCreditCard(int cardNumber, string holderName, int expiryDate, int startBalance, string supportAssistant) : base (cardNumber, holderName, expiryDate, startBalance)
    {
        SupportAssistant = supportAssistant;
    }

    public override void CheckBalance() 
    {
        if (_balance < 0) 
        {
            Console.WriteLine($"Баланс вашей кредитной карты отрицательный? Желаете получить помощь от {SupportAssistant}?");
        }
        else 
        {
            Console.WriteLine($"Баланс кредитной карты: {_balance} рублей");
        }
    }
}

public class CorporateCreditCard : CreditCard, ICorporateCreditCard
{
    public string Company { get; set; }

    public CorporateCreditCard(int cardNumber, string holderName, int expiryDate, int startBalance, string company) : base (cardNumber, holderName, expiryDate, startBalance)
    {
        Company = company;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Номер кредитной карты: {CardNumber}; Владелец: {HolderName}; Срок действия: {ExpiryDate}; Привязана к компании: {Company}.");
    }

}

public class Client 
{
    public string Name { get; set; }
    private ICreditCard _card;

    public Client(string name, ICreditCard card)
    {
        Name = name;
        _card = card;
    }

    public void Info() 
    {
        Console.WriteLine($"{Name} подал запрос на получение данных о карте:");
        _card.GetInfo();
    }

    public void Buy(int amount)
    {
        Console.WriteLine($"{Name} совершает покупку:");
        _card.Pay(amount);  
    }

    public void CheckMyBalance()
    {
        Console.WriteLine($"{Name} проверяет баланс:");
        _card.CheckBalance();
    }
}

GoldCreditCard gold = new GoldCreditCard(666666, "Гоголь-Моголь", 1025, 26700, 8);
Client client_number_one = new Client("Гоголь-Моголь", gold);
client_number_one.Info();
client_number_one.Buy(15000);         
client_number_one.CheckMyBalance();

Console.WriteLine("-----------------------------------------------------------------------------------------------------------------");

PremiumCreditCard premium = new PremiumCreditCard(545454, "Кузя", 0527, -100, "Создателя");
Client client_number_two = new Client("Кузя", premium);
client_number_two.Info();
client_number_two.Buy(1);         
client_number_two.CheckMyBalance();

Console.WriteLine("-----------------------------------------------------------------------------------------------------------------");

CorporateCreditCard corporate = new CorporateCreditCard(777777, "Корнилов", 0831, 999999, "ООО ТмывДенег");
Client client_number_tree = new Client("Корнилов", corporate);
client_number_tree.Info();
client_number_tree.Buy(999999);         
client_number_tree.CheckMyBalance();



```
